```python
class Father: #parent class
    def Walk(self):
        return "i walk"

    def Sing(self):
        return "i sang"


class Son(Father): #child class ## single level inheritance
    def Jump(self):
        return "i jump"


class Grandson(Son): ### multilevel inheritance
    def eat(self):
        return "i ate"
```

```python
user1 = Father()
```

```python
user1.Sing()
```

```python
user1.Walk()
```

```python
dir(user1)
```

```python
user2 = Son()
```

```python
dir(user2)
```

```python
user3 = Grandson()
```

```python
dir(user3)
```

```python
class Father:
    def Walk(self):
        return "father walk"
```

```python
class Son(Father):
    def Walk(self): ## method overriding
        return "son walk"
        
    def draw(self):
        return "son drew"
```

```python
user5 = Son()
```

```python
user5.draw()
```

```python
user5.Walk()
```
