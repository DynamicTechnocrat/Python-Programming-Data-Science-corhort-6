```python
import time
x = [1,3,5,7]


for a in x: #a = x[0] 1 , a = x[1] 3 , a = x[2] 5 , a = x[3] 7
    print(a + 2)
    time.sleep(2)
```

```python
a = x[0]
print(a + 2)
time.sleep(2)

a = x[1]
print(a + 2)
time.sleep(2)

a = x[2]
print(a + 2)
time.sleep(2)


a = x[3]
print(a + 2)
time.sleep(2)
```

```python
names = ['deji',"ayo","bola"]

for a in names:
    print(a.upper(),type(a))
    time.sleep(2)
```

```python
name2 = ["bola","tiffe","bello"]

for it_developer in name2:
    print(it_developer)
    time.sleep(2)
```

```python
name3 = ["fola","teju","bade"]

for nam in name3:
    if nam == "teju":
        print(f"i have seen {nam}")
    else:
        print(nam)
```

```python
name4 = ["male","female","female"]

for x in name4:
    print(x)
```

```python
## in order to print the index with the element use enumerate
name4 = ["male","female","female"]

for x in enumerate(name4):
    print(x)
```

```python
name5 = ("mrs ade" ,"mr james" ,"mr emmnauel")

for value in name5:
    print(value.title())
```

```python
name6 = ["mr's ade" , "mr bello" , "mr philip"]

for x in name6:
    print(f"here is {x} food")
```
