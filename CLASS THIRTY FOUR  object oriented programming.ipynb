```python
class Customer:
    def __init__(self , name , age ): ## attribute
        self.myname  = name
        self.myage = age

    def Walk(self): ### method
        return f"{self.myname} is walking"

    def Run(self):
        return f"{self.myname} ran"
```

```python
class BankAccount:
    def __init__(self , accountname , balance ):
        self.myaccountname = accountname
        self.__mybalance = balance

    def Deposit(self ,  amount ):
        self.__mybalance = self.__mybalance + amount
        return f"{amount} has been deposited successfully and your new account balance is {self.__mybalance}"

    def Withdraw(self , amount):
        if amount <= self.__mybalance:
            self.__mybalance = self.__mybalance - amount
            return f"{amount} withdraw successful your remaining balance is {self.__mybalance}"
        else:
            return "insufficient fund"

    def Checkbalance(self):
        return f"your current balance is {self.__mybalance}"
```
