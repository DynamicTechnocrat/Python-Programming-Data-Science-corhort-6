```python
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Calculate the statistical anchors
q1 = df3['Insulin'].quantile(0.25)
q2 = df3['Insulin'].median()
q3 = df3['Insulin'].quantile(0.75)

# 2. Set aesthetic style
sns.set_theme(style="whitegrid")
plt.figure(figsize=(8, 10))

# 3. Create the boxplot with specific styling
ax = sns.boxplot(
    y=df3['Insulin'], 
    color="#A2D2FF",     # Soft professional blue
    width=0.3, 
    linewidth=2,
    fliersize=8,
    flierprops={"marker": "x", "markeredgecolor": "red"} # Clear outlier marking
)

# 4. Annotate Q1, Q2, and Q3
# Offset annotations to the right for clarity
annotations = [
    (q1, f"Q1 (25%): {q1:.1f}", "#1B4965"),
    (q2, f"Q2 (Median): {q2:.1f}", "#D62828"), # Highlight median in red
    (q3, f"Q3 (75%): {q3:.1f}", "#1B4965")
]

for val, label, color in annotations:
    ax.text(0.22, val, label, va='center', weight='semibold', color=color)
    # Add a small guide line from the box to the text
    ax.hlines(y=val, xmin=0.15, xmax=0.21, colors=color, linestyles='dotted', lw=1)

# 5. Final Polishing
plt.title('Insulin Level Distribution & Quartile Tracking', fontsize=18, pad=20, fontweight='bold')
plt.ylabel('Insulin ($\mu U/mL$)', fontsize=14)
sns.despine(left=True, bottom=True) # Remove unnecessary chart borders

plt.show();
```
