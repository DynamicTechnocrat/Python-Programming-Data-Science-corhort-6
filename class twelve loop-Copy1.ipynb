```python
import time
x = [1,2,3,4]
```

```python
for num in x:
    print(num)
    time.sleep(2)
```

```python
for num2 in x:
   print(num2 * 2) ## 2 , 4
```

```python
for num3 in x:
    compute = num3 * 2
    print(compute)
    time.sleep(2)
```

```python
x = [1,2,3,4] 
y = []
for num3 in x:
    compute = num3 * 2 #2 4
    y.append(compute) #2 , 4
    print(compute) ### 2 has been added 4 has been added
    time.sleep(3)
```

```python
numbers = [1,2,3]

w = 0


w = w + numbers[0] # 0 + 1 = 1
print(w)
time.sleep(2)
w = w + numbers[1] # 1 + 2 = 3
print(w)
time.sleep(2)
w = w + numbers[2] # 3 + 3 = 6
print(w)
time.sleep(2)
```

```python
numbers = [1,2,3]
w = 0
for num in numbers:
    w = w + num # 0 + 1 = 1, 1+2 = 3 , 3+3 = 6
    print(w)
    time.sleep(2)
```

```python
number2 = [10,20,40]
p = 0

for num in number2:
    p = p + num
    print(p)
    time.sleep(2)
```
