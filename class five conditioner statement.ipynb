```python
gender = input("what is your gender : ")


if gender == "female":
    print("you are a woman")

else:
    print("you are a man")
```

```python
gender = input("what is your gender : ")


if gender.lower() == "female":
    print("you are a woman")

else:
    print("you are a man")
```

```python
gender = input("what is your gender : ")


if gender.lower().strip() == "female":
    print("you are a woman")

else:
    print("you are a man")
```
