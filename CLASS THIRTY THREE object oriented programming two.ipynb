```python
### CUSTOMER CLASS
## NAME AND AGE

## WALK , RUN
```

```python
class Customer:
    def __init__(self,name , age):
        self.myname = name
        self.myage = age

    def Walk(self): ### method 
        return f"{self.myname} walk"

    def Run(self):  ### method
        return f"{self.myname} ran"
```

```python
user1 = Customer("deji",12)
```

```python
dir(user1)
```

```python
user1.myname
```

```python
user1.myage
```

```python
user1.Walk()
```

```python
user1.Run()
```

```python
class EcommerceCustomer:
    def __init__(self , name , age , address):
        self.customer_name =  name 
        self.customer_age = age
        self.customer_address = address

    def PlaceOrder(self):
        return f"{self.customer_name} order placed"

    def WithdrawOrder(self):
        return f"{self.customer_name} order withdraw"
```
