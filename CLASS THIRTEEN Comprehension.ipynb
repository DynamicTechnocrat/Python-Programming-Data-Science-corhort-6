```python
import time
names = ["deji","ayo","bola"]

for name in names:
    print(name)
    time.sleep(2)
```

```python
import time
names = ["deji","ayo","bola"]

for name in names:
    print(name,type(name))
    time.sleep(2)
```

```python
import time
names = ["deji","ayo","bola"]

for name in names:
    print(name.upper())
    time.sleep(2)
```

```python
import time
names = ["deji","ayo","bola"]
name3 = []
for name in names:
    name3.append(name.upper())
    print(name.upper(),"append")
    time.sleep(2)
```

```python
name3
```

```python
names = ["deji","ayo","bola"]
name3 = [name.upper() for name in names]

name3
```
